```python
import matplotlib.pyplot as plt

models = [
    "Wavelet+NN",
    "LSTM",
    "LSTM-BiDir",
    "FCN",
    "ResNet",
    "Inception1D",
    "xResNet1D101",
]

auc = [
    0.849,
    0.907,
    0.914,
    0.918,
    0.919,
    0.925,
    0.925,
]

plt.figure(figsize=(10,6))

bars = plt.barh(models, auc)

for bar, value in zip(bars, auc):
    plt.text(
        value + 0.002,
        bar.get_y() + bar.get_height()/2,
        f"{value:.3f}",
        va="center"
    )

plt.xlim(0.80, 0.95)

plt.xlabel("Macro AUC")
plt.ylabel("Model")
plt.title("PTB-XL Benchmark Comparison")

plt.tight_layout()

plt.savefig(
    "ecg_model_selection.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
